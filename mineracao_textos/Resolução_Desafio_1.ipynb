````python
# importando as bibliotecas 
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from bs4 import BeautifulSoup
import time
````

````python
try:
    driver = webdriver.Chrome()
    # abrindo a página
    driver.get('https://pt.wikipedia.org/')
    driver.maximize_window()
    time.sleep(5)
    
    # selecionando a barra de busca
    search_bar = driver.find_element(By.NAME, "search")
    # digitando a palavra cesar
    search_bar.send_keys("cesar")

    time.sleep(10)
    
    # atualiza o botão de busca
    search_bar = driver.find_element(By.NAME, "search")
    #pressiona enter na pesquisa
    search_bar.send_keys(Keys.RETURN)
    
    #aguardando o retorno da página
    time.sleep(20)
    # clicando no link do cesar school
    link_cesar = driver.find_element(By.CSS_SELECTOR,'#mw-content-text > div.mw-content-ltr.mw-parser-output > ul:nth-child(12) > li:nth-child(2) > a')
    link_cesar.click()
    time.sleep(20)
    html = driver.page_source
    #fechando o navegador
    driver.quit()
except Exception as erro:
    print(f"Não consegui acessar a página - {erro}")
    driver.quit()
````

````python
#html
````

````python
# capturando os dados com o beautifulsoup
soup = BeautifulSoup(html, "html.parser")
````

````python
# capturando elementos das tags tr e td
tr_elements = soup.find_all("tr")

elementos_chave = {}

for row in tr_elements:
    cells = row.find_all("td")
    cell_texts = [cell.get_text(strip=True) for cell in cells]
    if len(cell_texts) == 2:
        chave = cell_texts[0]
        valor = cell_texts[1]

        elementos_chave[chave] = valor

print(elementos_chave)        

# tipo
# data fundacao
# fundador 1
# fundador 2
# fundador 3
# local da fundacao
# ceo
# coo
# numero funcionarios
# site
````

````python
# graducao_01
# graduacao_02
# mestrado_01 
# mestrado_02 
# doutorado_02

tabela = soup.find("table", class_="wikitable")

cursos = {'curso':[],'nome_curso':[]}

for t in tabela:
    tr_elementos = tabela.find_all('tr')
    th_elementos = tabela.find_all('th')
    td_elementos = tabela.find_all('td')

    for th in th_elementos:
        # print(f"th - {th.get_text(strip=True)}")
        curso = th.get_text(strip=True)
        if curso not in {'Cidade','Curso/Programa','ref'} and curso not in cursos['curso']:
            cursos['curso'].append(curso)

    for td in td_elementos:
        # print(f"td {td.get_text(strip=True)}")
        nome_curso = td.get_text(strip=True)
        if nome_curso not in {'[11]','Recife','RecifeeManaus','[8][15]','[4][14]','[16][15]'}:
            cursos['nome_curso'].append(nome_curso)

curso_ = {'grad':[],'mest':[],'dout':[]}

for i in range(0,5):
    if i <= 1:
        curso_['grad'].append(cursos['nome_curso'][i])
    elif i > 1 and i <=3:
        curso_['mest'].append(cursos['nome_curso'][i])
    elif i == 4:
        curso_['dout'].append(cursos['nome_curso'][i])

curso_
````

````python
# ESSA CÉLULA DEVE SER A ULTIMA DO SEU CÓDIGO E NENHUMA STRING DEVE SER COLOCADA NAS VARIÁVEIS DE FORMA MANUAL!
# NADA DE tipo = "INSTITUTO CESAR" POR EX. A IDEIA É PEGAR DA WIKI.

tipo                = elementos_chave['Tipo']
data_fundacao       = elementos_chave['Fundação']
fundador_01         = elementos_chave['Fundadores'].split(',')[0]
fundador_02         = elementos_chave['Fundadores'].split(',')[1].split(' e ')[0]
fundador_03         = elementos_chave['Fundadores'].split(',')[1].split(' e ')[-1]
local_fundacao      = elementos_chave['Sede']
ceo                 = elementos_chave['Pessoas importantes'].split('(CEO)')[0]
coo                 = elementos_chave['Pessoas importantes'].split('(CEO)')[1].split('(COO)')[0]
numero_funcionarios = elementos_chave['Empregados'][:4]
graduacao_01        = curso_['grad'][0]
graduacao_02        = curso_['grad'][1]
metrado_01          = curso_['mest'][0]
mestrado_01         = curso_['mest'][1]
doutorado_01        = curso_['dout'][0]
site                = elementos_chave['Sítio oficial']

print(f"""
      O CESAR é um {tipo} fundado em {data_fundacao} por {fundador_01}, {fundador_02}, {fundador_03}
      em {local_fundacao}. Segundo os dados coletados, {ceo} (CEO) e {coo} (COO) são os chefes da organização
      que tem {numero_funcionarios}. O CESAR possui cursos de {graduacao_01}, {graduacao_02}, {metrado_01},
      {mestrado_01}, {doutorado_01}. Para mais detalhes dessa instituição incrível, acesse {site}
""")
````
