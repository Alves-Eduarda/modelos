```python
import pandas as pd
import numpy as np
from sklearn.datasets import fetch_20newsgroups
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from sklearn.feature_extraction.text import TfidfTransformer
```

```python
# Descomente a linha de baixo se quiser ver mais detalhes do dataset
print(fetch_20newsgroups.__doc__)
```

```python
# selecionando a quantidade de textos que serão utilizados no desafio
N = 15
raw_documents= fetch_20newsgroups(subset='train', return_X_y=True)[0]

corpus = []

for idx, doc in enumerate(raw_documents[:N]):
  print(f"========== Doc {idx} ==========")
  print(doc)
  corpus.append(doc)
```

```python
len(corpus)
```

Observações:
* cada email é um documento
* o corpus vai ser formado pelos 15 documentos

###  Passos para serem implementados

* realizar processo da tokenização
* retirar palavras sem relevância com o stopword
* vetorização
* extração de características TF, IDF, TF - IDF


```python
nltk.download('punkt_tab')
nltk.download('stopwords')

stop_words = stopwords.words('english')

doc_tokens = []

for doc in corpus:
  tokenizacao = [token.lower() for token in word_tokenize(doc, language= 'english') if token.lower() not in stop_words]
  doc_tokens.append(tokenizacao)
```

```python
doc_tokens[0][:10]
```

```python
# transformando as palavras em sentenças
text_line = [' '.join(doc) for doc in doc_tokens]
text_line[0]
```

**Vetorização**

```python
# vetorização
from sklearn.feature_extraction.text import CountVectorizer

# retorna um vetor aonde cada linha representa um documento, cada coluna representa um termo e os valores são quantas vezes cada palavra aparece no respectivo documento
# ele tokeniza o texto quebrando em palavras

count_vectors = CountVectorizer(stop_words=None,min_df=0.005,max_df=0.5,max_features=3000)
count_vectors.fit(text_line)
print(f"Quantidade de documentos: {len(count_vectors.vocabulary_)} e Quantidade de palavras: {sum([doc[1] for doc in count_vectors.vocabulary_.items()])}")
```

```python
sorted(count_vectors.vocabulary_.items(),reverse=False)
```

```python
X = count_vectors.transform(text_line)
print(X[0,:])
```

```python
X.shape
```

```python
# visualizando o token
count_vectors.get_feature_names_out()[20]
```

```python
df_x = pd.DataFrame(X.toarray(),columns=count_vectors.get_feature_names_out()).T

df_x.reset_index(inplace=True)

df_x[df_x['index'] == '15']
```

### TF

```python
tf_transformer = TfidfTransformer(use_idf=False) #use_idf = False desabilita o processo do idf
X_tf = tf_transformer.fit_transform(X)

# Mostrar os resultados
feature_names = count_vectors.get_feature_names_out()
tf_array = X_tf.toarray()

print(X_tf[0,:])
```

```python
df_x = pd.DataFrame(tf_array,columns=count_vectors.get_feature_names_out()).T

df_x.reset_index(inplace=True)

df_x[df_x['index'] == '15']
```

```python
# Exibindo TF para cada documento
tf_values = {'documento':[],'palavras': [], 'TF':[]}

for i, doc in enumerate(tf_array):
    for palavra, tf in zip(feature_names, doc):
        if tf > 0:
            tf_values['palavras'].append(palavra)
            tf_values['TF'].append(tf)
            tf_values['documento'].append(i)


df_tf = pd.DataFrame(tf_values)
df_tf.head()
```

```python
df_tf[df_tf['palavras'] == '15']
```

### IDF

```python
# calcular IDF
idf_transformer = TfidfTransformer()
idf_transformer.fit(X)

# Acessando os valores de IDF
idf_valores = idf_transformer.idf_
palavras = count_vectors.get_feature_names_out()

idf_valores
```

```python
# valores do IDF
idf_values = {'palavras': [], 'IDF':[]}

for doc_index, doc_vector in enumerate(X):
    doc_terms = doc_vector.toarray()[0]
    for i, count in enumerate(doc_terms):
        if count > 0:
            idf_values['palavras'].append(palavras[i])
            idf_values['IDF'].append(idf_valores[i])

df_idf = pd.DataFrame(idf_values)
df_idf.head()
```

```python
df_idf[df_idf['palavras'] == '15']
```

### TF-IDF

calculado através do countVectorizer

```python
# calculando o TF-IDF

tfidf_transformer = TfidfTransformer()
tfidf_transformer.fit(X)

X_tfidf = tfidf_transformer.transform(X)
x_tfidf_array = X_tfidf.toarray()

# Mostrar os resultados
feature_names_tfidf = count_vectors.get_feature_names_out()

print(X_tfidf[0,:])
```

```python
tfidf_values = {'documento':[],'palavras':[],'TFIDF':[]}

for i, doc in enumerate(x_tfidf_array):
    for palavra, tf in zip(feature_names_tfidf, doc):
        if tf > 0:
            tfidf_values['palavras'].append(palavra)
            tfidf_values['TFIDF'].append(tf)
            tfidf_values['documento'].append(i)

df_tfidf = pd.DataFrame(tfidf_values)
df_tfidf.head()
```

```python
df_tfidf[df_tfidf['palavras'] == '15']
```

**Datframe Resultado**

```python
df_resultado = pd.merge(df_tf,df_idf,on='palavras',how='inner')
df_resultado = pd.merge(df_resultado,df_tfidf,on=['palavras','documento'],how='inner')

df_resultado.drop_duplicates(inplace=True)

df_resultado.head()
```

```python
df_resultado[df_resultado['palavras'] == '15']
```

Ordenando os casos com TFIDF de forma descrescente

```python
df_resultado.sort_values(by='TFIDF',ascending=False,inplace=True)
df_resultado.head()
#
```

Os 5 termos com maior TF-IDF significam que eles apresentam maior relevância sobre os seus respectivos documentos. Cada termo destacado se repete bastante no documento e faz parte do contexto que explica o texto. Exemplo:


* O termo **scsi** aparece bastante no documento zero e faz parte do contexto do texto que está conectado com a padronização de comunicação entre interfaces como os computadores.
* O termo **car** para o 2º maior TFIDF apresenta coerência com o texto pois o contexto do email se refere a um carro.
